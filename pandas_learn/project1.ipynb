```python
def e(txt):
    a = txt.find('(')
    b = txt.find(')')
    c = ''
    for i in txt[a+1:b]:
        if i.isdigit():
            c += i
    return c
            
```

```python
ep = []
for i in df['Title']:
    ep.append(int(e(i)))

df['Episodes'] = ep
df
```

```python
def ex_t(txt):
    a = txt.find(')')
    return txt[a+1:a+20]   
```

```python
time_stamp = []
for i in df['Title']:
    time_stamp.append(ex_t(i))

df['Time Stamp'] = time_stamp
df
```

```python
# Split into start and end dates
dates = df['Time Stamp'].str.split(' - ', expand=True)

# Convert to datetime
start = pd.to_datetime(dates[0], format='%b %Y')
end = pd.to_datetime(dates[1], format='%b %Y')

# Calculate months
df['Months'] = (end.dt.year - start.dt.year) * 12 + (end.dt.month - start.dt.month)

df
```
