# Pandas Series
A series is one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(arr,index = labels)
```

```python
pd.Series(d)
```

# Creating DataFrame

```python
data = {
    "Name" : ['Himanshu','Bhavik','Ankit','Parnay'],
    "Stream" : ['non medical','medical','non medical','arts'],
    'age' : [18,19,20,18],
    'score' : [94, 85, 80, 97]
}
labels = [1,2,3,4]
```
