#### Correlation describes how strongly two variables tend to change together in a linear relationship.

#### Its value lies between -1 and +1.

Covariance = summation(Xi-X)(Yi-Y)/(n-1)
Correlation = Covariance/std(x)std(y)
if cor is close to 1 : it is highly correlates
if cor is close to 0: it has no correlation
if cor is close to -1 : it is negatively correlates
