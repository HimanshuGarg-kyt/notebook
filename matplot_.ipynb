```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,1,20)
y = x**2
```

```python
plt.plot(x,y,color = "pink")
plt.title("graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

```python
plt.subplot(1,2,1)
plt.plot(x**3,x)
plt.subplot(1,2,2)
plt.plot(x,y)
plt.title("graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

```python
a = plt.figure(figsize = (9,6),dpi = 150,facecolor="yellow",edgecolor="red")
a1 = a.add_axes([0,0,0.5,0.5])
plt.plot(x,y)
plt.title("graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
a2 = a.add_axes([0.6,0,0.5,0.5])
plt.plot(x,y**3)
plt.title("graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```
