```python
batsman = pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'green',linestyle = 'dashed',linewidth=3,marker = '<',markersize = 10,label = 'Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'red',linestyle = 'dashdot',marker = 'D',markersize = 10,label = 'rohit sharma')
plt.title('Comparison of rohit sharma and Virat kholi')
plt.xlabel('year')
plt.ylabel('Score')
plt.legend()
plt.grid()
# line style = dotted,solid,dashed , dashdot
# marker = D, o, +, ., >, <
# loc = best(default), upper right, upper left,lower left,lower right, center right etc.
```

```python
np.random.seed(40)
price = np.random.randint(40000,100000,6)
price = np.append(price,10000000)
year = np.arange(2014,2021)
plt.plot(year,price)
plt.ylim(40000,100000)  # limit the y axis 
plt.grid()
plt.show()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'green',linestyle = 'dashed',linewidth=3,marker = '<',markersize = 10,label = 'virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'red',linestyle = 'dashdot',marker = 'D',markersize = 10,label = 'rohit sharma')
plt.title('Comparison of rohit sharma and Virat kholi')
plt.xlabel('year')
plt.ylabel('Score')
plt.legend(loc='lower right')
plt.grid()
plt.show()
np.random.seed(40)
price = np.random.randint(40000,100000,6)
price = np.append(price,10000000)
year = np.arange(2014,2021)
plt.plot(year,price)
plt.ylim(40000,100000)  # limit the y axis 
plt.grid()
plt.show()
```

```python
plt.bar(df1['batsman'],df1['2015'],color = 'red',label = '2015')
plt.bar(df1['batsman'],df1['2016'],bottom=df1['2015'],color = 'yellow',label = '2016')
plt.bar(df1['batsman'],df1['2017'],bottom=df1['2016']+df1['2015'],color = 'blue',label = '2017')
plt.legend()
```
